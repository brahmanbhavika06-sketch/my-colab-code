```python
from langchain_groq import ChatGroq
from langchain_mistralai import ChatMistralAI
import os

# Groq model
groq_llm = ChatGroq(
    model="llama-3.1-8b-instant",
    api_key=os.getenv("GROQ_API_KEY")
)

# Mistral model
mistral_llm = ChatMistralAI(
    model="mistral-small-2603",
    api_key=os.getenv("MISTRAL_API_KEY")
)

# Same question for both models
question = "What are the advantages of using LangChain?"

# Groq response
groq_response = groq_llm.invoke(question)

# Mistral response
mistral_response = mistral_llm.invoke(question)

# Print both responses
print("========== Groq Response ==========")
print(groq_response.content)

print("\n========== Mistral Response ==========")
print(mistral_response.content)
```

```python
from langchain_groq import ChatGroq
import os

prompt = "Write a short creative story about a robot learning to cook."

temperatures = [0.1, 0.7, 1.2]

for temp in temperatures:
    print("\n" + "=" * 60)
    print(f"Temperature: {temp}")
    print("=" * 60)

    llm = ChatGroq(
        model="llama-3.1-8b-instant",
        temperature=temp,
        api_key=os.getenv("GROQ_API_KEY")
    )

    response = llm.invoke(prompt)

    print(response.content)
```

```python
from langchain_groq import ChatGroq
import os
llm = ChatGroq(
    model="llama-3.1-8b-instant",
    api_key=os.getenv("GROQ_API_KEY")
)

def structured_prompt(topic):
    prompt = f"""
You are an AI tutor.

Explain the topic "{topic}" using the following structure:

1. Short Definition:
Give a simple and clear definition in 2-3 sentences.

2. Three Key Points:
- Key point 1
- Key point 2
- Key point 3

3. One Real-Life Example:
Give one simple real-life example.

Keep the explanation easy to understand.
"""

    response = llm.invoke(prompt)
    return response.content

# Test the function
result = structured_prompt("Machine Learning")

print(result)
```

```python
from langchain_groq import ChatGroq
from langchain_mistralai import ChatMistralAI
import os

# Check API keys
groq_key = os.getenv("GROQ_API_KEY")
mistral_key = os.getenv("MISTRAL_API_KEY")

if not groq_key:
    print("Warning: GROQ_API_KEY is missing.")

if not mistral_key:
    print("Warning: MISTRAL_API_KEY is missing.")


# Create models
groq_llm = None
mistral_llm = None

if groq_key:
    groq_llm = ChatGroq(
        model="llama-3.1-8b-instant",
        api_key=groq_key
    )

if mistral_key:
    mistral_llm = ChatMistralAI(
        model="mistral-small-2603",
        api_key=mistral_key
    )


print("🤖 Multi-Model AI Assistant")
print("Choose Groq or Mistral.")
print("Type 'quit' to exit.")

while True:

    choice = input("\nChoose model (Groq/Mistral): ")

    if choice.lower() == "quit":
        print("Assistant: Goodbye! 👋")
        break

    if choice.lower() not in ["groq", "mistral"]:
        print("Please choose either Groq or Mistral.")
        continue

    question = input("Enter your question: ")

    if question.lower() == "quit":
        print("Assistant: Goodbye! 👋")
        break

    try:

        if choice.lower() == "groq":

            if groq_llm is None:
                print("Error: Groq API key is missing.")
                continue

            response = groq_llm.invoke(question)

        else:

            if mistral_llm is None:
                print("Error: Mistral API key is missing.")
                continue

            response = mistral_llm.invoke(question)

        print("\nAssistant Response:")
        print(response.content)

    except Exception as e:
        print("\nError: Model failed.")
        print("Details:", e)
```
