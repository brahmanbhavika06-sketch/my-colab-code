```python
# Q1. Linear Regression

# Import Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Step 1: Load Dataset
df_lr = pd.read_csv("ford_car_dataset.csv")

print("Dataset Loaded Successfully!")

print("\nFirst 5 Rows:")
print(df_lr.head())

print("\nDataset Shape:")
print(df_lr.shape)

# Step 2: Data Cleaning

# Remove duplicate rows
df_lr = df_lr.drop_duplicates()

# Remove missing values
df_lr = df_lr.dropna()

print("\nAfter Data Cleaning:")
print("Shape:", df_lr.shape)

# Step 3: Encode Categorical Columns

df_lr = pd.get_dummies(df_lr, drop_first=True)

# Step 4: Feature and Target

X_lr = df_lr.drop("price", axis=1)
y_lr = df_lr["price"]

print("\nFeature Shape:", X_lr.shape)
print("Target Shape:", y_lr.shape)

# Step 5: Train-Test Split

X_train_lr, X_test_lr, y_train_lr, y_test_lr = train_test_split(
    X_lr,
    y_lr,
    test_size=0.20,
    random_state=42
)

# Step 6: Feature Scaling

scaler_lr = StandardScaler()

X_train_lr = scaler_lr.fit_transform(X_train_lr)
X_test_lr = scaler_lr.transform(X_test_lr)

print("\nFeature Scaling Completed!")

# Step 7: Train Linear Regression Model

lr_model = LinearRegression()

lr_model.fit(X_train_lr, y_train_lr)

print("\nLinear Regression Model Trained Successfully!")

# Step 8: Prediction

y_pred_lr = lr_model.predict(X_test_lr)

# Step 9: R2 Score

r2_lr = r2_score(y_test_lr, y_pred_lr)

print("\nR2 Score:", r2_lr)

# Step 10: Actual vs Predicted Prices

result_lr = pd.DataFrame({
    "Actual Price": y_test_lr.values,
    "Predicted Price": y_pred_lr
})

print("\nActual vs Predicted Price:")
print(result_lr.head(10))
```

```python
# Q2. Logistic Regression

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Step 1: Load Dataset
df_log = pd.read_csv("heart.csv")

print("Dataset Loaded Successfully!")

# Step 2: Data Cleaning
df_log = df_log.drop_duplicates()
df_log = df_log.dropna()

# Step 3: Encode Categorical Columns
df_log = pd.get_dummies(df_log, drop_first=True)

# Step 4: Feature and Target
X_log = df_log.drop("HeartDisease", axis=1)
y_log = df_log["HeartDisease"]

# Step 5: Train-Test Split
X_train_log, X_test_log, y_train_log, y_test_log = train_test_split(
    X_log,
    y_log,
    test_size=0.20,
    random_state=42
)

# Step 6: Feature Scaling
scaler_log = StandardScaler()

X_train_log = scaler_log.fit_transform(X_train_log)
X_test_log = scaler_log.transform(X_test_log)

# Step 7: Train Logistic Regression Model
log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train_log, y_train_log)

# Step 8: Prediction
y_pred_log = log_model.predict(X_test_log)

# Step 9: Evaluation
cm = confusion_matrix(y_test_log, y_pred_log)
accuracy = accuracy_score(y_test_log, y_pred_log)
precision = precision_score(y_test_log, y_pred_log)
recall = recall_score(y_test_log, y_pred_log)
f1 = f1_score(y_test_log, y_pred_log)

# Step 10: Print Results
print("Confusion Matrix:")
print(cm)

print("\nAccuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```

```python
# Q3. K-Nearest Neighbors (KNN)

# Import Libraries
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Dictionary to store accuracy for different k values
knn_results = {}

# Try different values of k
for k in [3, 5, 7]:

    # Create KNN model
    knn_model = KNeighborsClassifier(n_neighbors=k)

    # Train model
    knn_model.fit(X_train_log, y_train_log)

    # Prediction
    y_pred_knn = knn_model.predict(X_test_log)

    # Calculate accuracy
    accuracy_knn = accuracy_score(y_test_log, y_pred_knn)

    # Store result
    knn_results[k] = accuracy_knn

    print("k =", k, "Accuracy =", accuracy_knn)

# Find best k
best_k = max(knn_results, key=knn_results.get)

print("\nBest k value:", best_k)
print("Best Accuracy:", knn_results[best_k])
```

```python
# Q4. Naive Bayes

# Import Libraries
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, confusion_matrix

# Create Naive Bayes model
nb_model = GaussianNB()

# Train the model
nb_model.fit(X_train_log, y_train_log)

print("Naive Bayes Model Trained Successfully!")

# Make predictions
y_pred_nb = nb_model.predict(X_test_log)

# Confusion Matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_log, y_pred_nb))

# Classification Report
print("\nClassification Report:")
print(classification_report(y_test_log, y_pred_nb))
```

```python
# Q5. Algorithm Comparison

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd


# 1. Logistic Regression Metrics

accuracy_log = accuracy_score(y_test_log, y_pred_log)
precision_log = precision_score(y_test_log, y_pred_log)
recall_log = recall_score(y_test_log, y_pred_log)
f1_log = f1_score(y_test_log, y_pred_log)


# 2. Best KNN Model

# Train KNN using the best k found in Q3
best_knn_model = KNeighborsClassifier(n_neighbors=best_k)

best_knn_model.fit(X_train_log, y_train_log)

y_pred_best_knn = best_knn_model.predict(X_test_log)

accuracy_knn = accuracy_score(y_test_log, y_pred_best_knn)
precision_knn = precision_score(y_test_log, y_pred_best_knn)
recall_knn = recall_score(y_test_log, y_pred_best_knn)
f1_knn = f1_score(y_test_log, y_pred_best_knn)


# 3. Naive Bayes Metrics

accuracy_nb = accuracy_score(y_test_log, y_pred_nb)
precision_nb = precision_score(y_test_log, y_pred_nb)
recall_nb = recall_score(y_test_log, y_pred_nb)
f1_nb = f1_score(y_test_log, y_pred_nb)


# 4. Comparison Table
comparison_table = pd.DataFrame({
    "Algorithm": [
        "Logistic Regression",
        "KNN",
        "Naive Bayes"
    ],
    "Accuracy": [
        accuracy_log,
        accuracy_knn,
        accuracy_nb
    ],
    "Precision": [
        precision_log,
        precision_knn,
        precision_nb
    ],
    "Recall": [
        recall_log,
        recall_knn,
        recall_nb
    ],
    "F1 Score": [
        f1_log,
        f1_knn,
        f1_nb
    ]
})

print("Algorithm Comparison:")
print(comparison_table.to_string(index=False))


# 5. Find Best Algorithm

best_algorithm = comparison_table.loc[
    comparison_table["F1 Score"].idxmax(),
    "Algorithm"
]

print("\nBest Performing Algorithm:", best_algorithm)
```
