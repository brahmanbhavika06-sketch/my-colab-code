```python
import pandas as pd

# Load train.txt
df = pd.read_csv(
    "train.txt",
    sep=";",
    header=None,
    names=["text", "emotion"]
)

print("Dataset Shape:", df.shape)
print(df.head())
```

```python
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

# Apply cleaning
df["cleaned_text"] = df["text"].apply(clean_text)

print(df[["text", "cleaned_text", "emotion"]].head())
```

```python
from sklearn.model_selection import train_test_split

# Separate features and target
X = df["cleaned_text"]
y = df["emotion"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Print shapes
print("X_train Shape:", X_train.shape)
print("X_test Shape:", X_test.shape)
print("y_train Shape:", y_train.shape)
print("y_test Shape:", y_test.shape)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Create CountVectorizer
bow_vectorizer = CountVectorizer()

# Fit and transform training data
X_train_bow = bow_vectorizer.fit_transform(X_train)

# Transform test data
X_test_bow = bow_vectorizer.transform(X_test)

# Print shapes
print("X_train_BOW Shape:", X_train_bow.shape)
print("X_test_BOW Shape:", X_test_bow.shape)

# Display first 20 feature names
print("\nFirst 20 Feature Names:")
print(bow_vectorizer.get_feature_names_out()[:20])
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Create model
nb_bow = MultinomialNB()

# Train model
nb_bow.fit(X_train_bow, y_train)

# Make predictions
y_pred_bow = nb_bow.predict(X_test_bow)

# Calculate accuracy
accuracy_bow = accuracy_score(y_test, y_pred_bow)

print("Bag of Words Accuracy:", accuracy_bow)
print("Bag of Words Accuracy (%):", accuracy_bow * 100)
```

```python
# Total vocabulary size
print("Total Vocabulary Size:", len(bow_vectorizer.vocabulary_))

# Show 15 vocabulary words
print("\n15 Vocabulary Words:")
print(list(bow_vectorizer.vocabulary_.keys())[:15])

# Select one sample document
sample_document = X_train.iloc[0]

print("\nSample Document:")
print(sample_document)

# Convert sample document into BoW vector
sample_vector = bow_vectorizer.transform([sample_document])

print("\nBoW Vector:")
print(sample_vector.toarray())
```

```python
# Create CountVectorizer with unigrams and bigrams
bigram_vectorizer = CountVectorizer(
    ngram_range=(1, 2)
)

# Fit and transform training data
X_train_bigram = bigram_vectorizer.fit_transform(X_train)

# Transform test data
X_test_bigram = bigram_vectorizer.transform(X_test)

# Print shapes
print("X_train Bigram Shape:", X_train_bigram.shape)
print("X_test Bigram Shape:", X_test_bigram.shape)

# Display some features
print("\nSome Unigram and Bigram Features:")
print(bigram_vectorizer.get_feature_names_out()[:30])
```

```python
# Create Multinomial Naive Bayes model
nb_bigram = MultinomialNB()

# Train model
nb_bigram.fit(X_train_bigram, y_train)

# Make predictions
y_pred_bigram = nb_bigram.predict(X_test_bigram)

# Calculate accuracy
accuracy_bigram = accuracy_score(y_test, y_pred_bigram)

print("Bigram Accuracy:", accuracy_bigram)
print("Bigram Accuracy (%):", accuracy_bigram * 100)

print("\nComparison:")
print("Unigram BoW Accuracy (%):", accuracy_bow * 100)
print("Unigram + Bigram Accuracy (%):", accuracy_bigram * 100)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Fit and transform training data
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)

# Transform test data
X_test_tfidf = tfidf_vectorizer.transform(X_test)

# Print shapes
print("X_train TF-IDF Shape:", X_train_tfidf.shape)
print("X_test TF-IDF Shape:", X_test_tfidf.shape)

# Display first 15 feature names
print("\nFirst 15 TF-IDF Feature Names:")
print(tfidf_vectorizer.get_feature_names_out()[:15])
```

```python
# Create Multinomial Naive Bayes model
nb_tfidf = MultinomialNB()

# Train model
nb_tfidf.fit(X_train_tfidf, y_train)

# Make predictions
y_pred_tfidf = nb_tfidf.predict(X_test_tfidf)

# Calculate accuracy
accuracy_tfidf = accuracy_score(y_test, y_pred_tfidf)

print("TF-IDF Accuracy:", accuracy_tfidf)
print("TF-IDF Accuracy (%):", accuracy_tfidf * 100)
```

```python
comparison = pd.DataFrame({
    "Method": [
        "Bag of Words (Unigrams)",
        "Bag of Words (Unigrams + Bigrams)",
        "TF-IDF"
    ],

    "Accuracy": [
        accuracy_bow,
        accuracy_bigram,
        accuracy_tfidf
    ],

    "Accuracy (%)": [
        accuracy_bow * 100,
        accuracy_bigram * 100,
        accuracy_tfidf * 100
    ]
})

display(comparison)

best_index = comparison["Accuracy"].idxmax()

print("Best Method:", comparison.loc[best_index, "Method"])
print("Best Accuracy (%):", comparison.loc[best_index, "Accuracy (%)"])
```

```python
import joblib

# Store accuracies
results = {
    "Bag of Words": accuracy_bow,
    "Bag of Words + Bigrams": accuracy_bigram,
    "TF-IDF": accuracy_tfidf
}

# Find best method
best_method = max(results, key=results.get)
best_accuracy = results[best_method]

print("Best Method:", best_method)
print("Best Accuracy:", best_accuracy)
print("Best Accuracy (%):", best_accuracy * 100)
```

```python
if best_method == "Bag of Words":
    best_model = nb_bow
    best_vectorizer = bow_vectorizer

elif best_method == "Bag of Words + Bigrams":
    best_model = nb_bigram
    best_vectorizer = bigram_vectorizer

else:
    best_model = nb_tfidf
    best_vectorizer = tfidf_vectorizer

# Save best model
joblib.dump(best_model, "best_emotion_model.pkl")

# Save best vectorizer
joblib.dump(best_vectorizer, "best_emotion_vectorizer.pkl")

print("Best model saved successfully!")
print("Best vectorizer saved successfully!")
```
